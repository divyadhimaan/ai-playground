```python
resume = "/Users/divyadhiman/Documents/Resume/sde2/fullstack/Divya Dhiman's Resume.pdf"
```

```python
import pdfplumber

def read_resume(file_path: str) -> str:
    """
    Extract text from resume PDF.
    Optimized for LLM ingestion.
    """
    text_chunks = []

    with pdfplumber.open(file_path) as pdf:
        for page in pdf.pages:
            content = page.extract_text()
            if content:
                text_chunks.append(content)

    return "\n".join(text_chunks)

def preprocess(text: str) -> str:
    """
    Clean text for better LLM performance.
    """
    return " ".join(text.split())
```

```python
resume_text = preprocess(read_resume(resume))
```

```python
# Step 1: Create your prompts

system_prompt = """
You are a LinkedIn message writer.

Your task is to write a referral request message FROM THE USER'S PERSPECTIVE.

Instructions:
- Write the message as if the user is sending it to someone on LinkedIn proving a right fit.
- Use first person ("Hi, I came across...", "I am currently...", etc.)
- Keep it concise (80–120 words)
- Add tech stack and reference of experience from resume
- Be professional, polite, and slightly warm
- Highlight relevant experience from the resume
- Align with the job description, dont add things not in resume
- End with a clear ask for referral

Do NOT describe the user. Write AS the user.
Do NOT add placeholders.
"""

user_prompt = f"""
    Here is the resume of the user:
{resume_text}

Here is the job description:
{job_description}

Write a short LinkedIn referral request message.
"""

# Step 2: Make the messages list

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
    ]

# Step 3: Call OpenAI
response = openai.chat.completions.create(
        model = "llama3.2",
        messages = messages
    )
# Step 4: print the result
print(response.choices[0].message.content)
```
